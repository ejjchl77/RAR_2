```python
import cv2


# 이미지를 불러옵니다.
img = cv2.imread('./data_test/test_crop/test2.jpg')


# Super Resolution을 하기 위해 2배 확대를 위한 ESPCN 모델을 사용합니다.
sr = cv2.dnn_superres.DnnSuperResImpl_create()
# 모델 파일의 이름을 적어주면 됩니다.
sr.readModel('ESPCN_x3.pb')
# 모델 파일에 적힌 숫자와 일치하도록 적어줘야 합니다.
sr.setModel("espcn",3)
# img를 입력 받아 결과로 result1를 돌려줍니다. 
result1 = sr.upsample(img)


# OpenCV에서 제공하는 보간법을 사용해봅니다.
result2 = cv2.resize(img, dsize=None, fx=2, fy=2)
cv2.imwrite('./data_test/test_crop/resol_test2.jpg', result2)
# 입력 이미지와 Super Resolution, OpenCV 보간법 적용 결과를 화면에 보여줍니다.
cv2.imshow('origianl', img)
cv2.imshow('result - ESPCN', result1)
cv2.imshow('result - Opencv', result2)
cv2.waitKey(0)
```

```python
import utils.lib_images_io as lib_images_io
import cv2

def select_images_loader(src_data_type, src_data_path):
    if src_data_type == "video":
        images_loader = lib_images_io.ReadFromVideo(
            src_data_path,
            sample_interval=SRC_VIDEO_SAMPLE_INTERVAL)

    elif src_data_type == "folder":
        images_loader = lib_images_io.ReadFromFolder(
            folder_path=src_data_path)

    elif src_data_type == "webcam":
        if src_data_path == "":
            webcam_idx = 0
        elif src_data_path.isdigit():
            webcam_idx = int(src_data_path)
        else:
            webcam_idx = src_data_path
        images_loader = lib_images_io.ReadFromWebcam(
            SRC_WEBCAM_MAX_FPS, webcam_idx)
    return images_loader

# -- Image reader and displayer
src_data_path = 'data_test/EJ_single_test_1106/train_data/'
images_loader = lib_images_io.ReadFromFolder(folder_path=src_data_path)
img_displayer = lib_images_io.ImageDisplayer()
img = images_loader.read_image()
img = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)
img_displayer.display(img, wait_key_ms=1)
```

```python
import pandas as pd
import os, glob
txt_file=glob.glob('/home/ejjchl77/RAR_2/output/test_02/skeletons/*.txt')
txt_file.sort()

# 1-100: reclining
# 101-200: seated
# 201-300: sleeping
# 301-400: walking

#error=0; acc =0 
activity=['reclining', 'seated', 'sleeping', 'walking']
reclining_acc = 0
reclining_err = 0
seated_acc = 0
seated_err = 0
sleeping_acc = 0
sleeping_err = 0
walking_acc = 0
walking_err = 0

for f in txt_file:
    f_base=os.path.basename(f)
    f_base=f_base.split('.')[0]
    
    num_file = int(f_base)
    error=0; acc =0
    if num_file >=1 and num_file < 101:
        out=pd.read_csv(f, header=None)
        try: 
            out_label=str(out.iloc[:,1][0]).replace('"', '')
            out_label=out_label.strip()

            if out_label == activity[0]:
                #print(f_base, out_label, '\n')
                acc=1
            else: error=1
        except: 
            error=1
                
        reclining_acc+=acc
        reclining_err+=error
    
        
    elif num_file >=101 and num_file < 201:
        out=pd.read_csv(f, header=None)
        try: 
            out_label=str(out.iloc[:,1][0]).replace('"', '')
            out_label=out_label.strip()

            if out_label == activity[1]:
                #print(f_base, out_label, '\n')
                acc=1
            else: error=1
        except: 
            error=1
            
        seated_acc+=acc
        seated_err+=error
        
    elif num_file >=201 and num_file < 301:
        out=pd.read_csv(f, header=None)
        try: 
            out_label=str(out.iloc[:,1][0]).replace('"', '')
            out_label=out_label.strip()

            if out_label == activity[2]:
                #print(f_base, out_label, '\n')
                acc=1
            else: error=1
        except: 
            error=1
            
        sleeping_acc+=acc
        sleeping_err+=error
        
    elif num_file >=301 and num_file < 401:
        out=pd.read_csv(f, header=None)
        try: 
            out_label=str(out.iloc[:,1][0]).replace('"', '')
            out_label=out_label.strip()

            if out_label == activity[3]:
                #print(f_base, out_label, '\n')
                acc=1
            else: error=1
        except: 
            error=1
            
        walking_acc+=acc
        walking_err+=error
        
        
        
        
print(acc)       
print('%s Accuracy: %d %%' %(activity[0], reclining_acc))
print('%s Accuracy: %d %%' %(activity[1], seated_acc))
print('%s Accuracy: %d %%' %(activity[2], sleeping_acc))
print('%s Accuracy: %d %%' %(activity[3], walking_acc))




```
