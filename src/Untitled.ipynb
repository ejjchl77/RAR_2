```python
from ast import literal_eval
import glob
from collections import Counter

def act2met(act):
    if act == 'sitting': met=1.0
    elif act == 'stand': met=1.2
    elif act == 'walking': met=1.7
    else: met = None  
    return met
    
def activity(out_files):
    id1=[]; id2=[]; id3=[]
    for out in out_files:
        
        f=open(out, 'r')
        lines=f.readlines()
        for line in lines:
            line=literal_eval(line)
            
            for ll in line:
                if not ll: activity.append(None)
                
                else: 
                    ids = ll[0][0]
                    act = ll[0][1]
                    
                    if ids == 1: id1.append(act2met(act))
                    elif ids == 2: id2.append(act2met(act))
                    elif ids == 3: id3.append(act2met(act))

    return id1, id2, id3

path = '/home/ejjchl77/RAR_2/data_test/mockup_3people/output/03-06-14-19-10-211/skeletons/'
path_ = glob.glob(path+'*.txt')
id1, id2, id3=activity(path_)
# id1=[i[0] for i in a]
# id2=[i[1] for i in a]
# id3=[i[2] for i in a]

print(id1, id2, id3)
```

```python
freq_id1=Counter(id1).most_common()[0][0]
freq_id2=Counter(id2).most_common()[0][0]
freq_id3=Counter(id3).most_common()[0][0]

print(freq_id1, freq_id2, freq_id3)
```
