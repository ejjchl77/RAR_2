```python
import os
import ast
activity = 'sitting'
# 1. 폴더 안에 있는 filename.txt 파일들을 하나씩 오픈하기
folder_path = "D:\\Mockup\\net\\RAR_2\\output\\EJ_single_test_1106\\"+ activity + "\\skeletons\\"  # 폴더 경로를 입력하세요.
file_list = os.listdir(folder_path)

acc = 0  # "walking"이 발견된 횟수를 저장할 변수 초기화

# 폴더 내의 모든 파일에 대해
for file in file_list:
    if file.endswith('.txt'):  # 파일이 .txt로 끝나는 경우에만 처리
        with open(os.path.join(folder_path, file), 'r') as f:
            lines = f.readlines()
            try:
                lines = ast.literal_eval(lines[0])
                lines = lines[0][0]

                if activity in lines[1]:
                    acc += 1
            except: pass
print("전체 데이터 수 : ", len(file_list))
print(f'"walking"이 발견된 횟수: {acc}')
print("acc:", acc/(len(file_list)-2)*100)
```
